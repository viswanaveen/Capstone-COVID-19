### High-Level Explanation for the Code Block

This code block creates a new DataFrame containing only the most recent data:

*   **`top = df[df['Date'] == '2020-07-27']`**: This line filters the main DataFrame (`df`) to select all rows where the 'Date' column is exactly '2020-07-27' (the latest date in the dataset) and stores these selected rows in a new DataFrame called `top`.
*   **`top`**: This command then displays the contents of this new `top` DataFrame, showing the data for the last day.

```python
# poulating rows with max date in top
top = df[df['Date'] == '2020-07-27']
top
```

 groups the `top` DataFrame (which contains data for the most recent date) by each 'Country'. For each country, it then calculates the total sum of 'Confirmed', 'Deaths', 'Recovered', and 'Active' cases. Finally, `reset_index()` converts the grouped result back into a standard table format.

```python
# Grouping all rows absed on country
top.groupby(by='Country')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
```

 groups the main DataFrame (`df`) by each unique 'Date'. For each date, it sums up all the 'Confirmed' cases from every entry. The `reset_index()` function then turns this grouped data back into a neat table named `confirmed`, where each row represents a date and its corresponding total confirmed cases.

```python
# Populating all confirmed case in confirmed dataset
confirmed = df.groupby(by='Date')['Confirmed'].sum().reset_index()
confirmed
```

 This creates a line graph. It uses your `confirmed` data, with 'Date' on the horizontal axis and 'Confirmed' cases on the vertical axis, showing how the number of confirmed cases changed day by day.
*   **`plt.xticks(rotation=91)`**: This rotates the date labels on the x-axis to prevent them from overlapping and makes them easier to read.
*   **`plt.show()`**: This command displays the generated line plot.

```python
# Visualizing confirmed case
plt.figure(figsize = (18,6))
sns.lineplot(data=confirmed, x='Date', y='Confirmed')
plt.xticks(rotation=91)
plt.show()
```

# Forecasting for confirmed cases

```python
# renaming confirmed columns to ds and y
confirmed.columns = ['ds','y']
confirmed.head()
```

```python
# importing prophet library
from prophet import Prophet
m = Prophet()
```

```python
m.fit(confirmed)
```

```python
# Making prediction for next 7 dates
future = m.make_future_dataframe(periods=7)

forecast = m.predict(future)
forecast.tail(9)
```

```python
forecast_next_7_days = forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(7)
forecast_next_7_days
```

```python
forecast[['ds','yhat','yhat_upper','yhat_lower']]
'''
ds:
This column contains the dates for which the predictions have been made. It stands for "date stamp."

yhat:
This is the predicted value (forecast) for the given date in the ds column. It represents the model's best estimate for that point in time.

yhat_upper:
This column represents the upper bound of the predicted value, which indicates the higher end of the confidence interval.
It suggests that the actual value is unlikely to exceed this amount, giving you an idea of the prediction's uncertainty.

yhat_lower:
This column represents the lower bound of the predicted value, indicating the lower end of the confidence interval. It suggests that the actual value is unlikely to be lower than this amount.
'''
```
